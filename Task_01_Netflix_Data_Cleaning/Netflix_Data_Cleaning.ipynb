# Netflix Data Cleaning and Preparation

## Project Overview

This project focuses on cleaning and preparing the Netflix Movies and TV Shows dataset using Python and Pandas.

The main objective is to inspect the dataset, identify data quality issues, standardize text columns, and export a cleaned dataset for further analysis.

- Python
- Pandas
- Jupyter Notebook

## 3. Data Quality Assessment

In this step, the dataset is checked for missing values and duplicate records.

- Missing values are identified using `isnull().sum()`.
- Duplicate records are identified using `duplicated().sum()`.

These checks help evaluate the quality and reliability of the dataset.

```python
df.isnull().sum()
```

## 4. Data Cleaning and Standardization

In this step, text columns are cleaned to remove unwanted leading and trailing whitespace.

- Country column is standardized using `str.strip()`.
- Type column is standardized using `str.strip()`.
- Rating column is inspected to verify its formatting.
- Unique values are examined to identify inconsistent entries.

The original information is preserved wherever no correction is required.

```python
df['country'].str.strip().equals(df['country'])
```

```python
df['country'].unique()
```

```python
df['country'].value_counts().head(10)
```

```python
(df['country'] == 'Not Given').sum()
```

```python
df['country'] = df['country'].str.strip()
```

```python
df['rating'].str.strip().equals(df['rating'])
```

```python
df['type'].value_counts()
```

```python
df['type'] = df['type'].str.strip()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDataset Shape:")
print(df.shape)
```

## 5. Exporting Cleaned Dataset

The cleaned dataset is exported as a CSV file using Pandas.

The output file is saved as `Netflix_Cleaned.csv` inside the output folder.

This cleaned dataset can be reused for further data analysis tasks.

```python
df.to_csv("output/Netflix_Cleaned.csv", index=False)
```
