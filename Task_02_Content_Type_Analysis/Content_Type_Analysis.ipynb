## 6. Content Ratings Analysis

The content ratings are analyzed to understand how Movies and TV Shows are distributed across different rating categories.

A comparison is performed using Seaborn's countplot to identify the most common ratings for each content type.

The rating categories are sorted by their frequency for better visualization.

```python
plt.figure(figsize=(14, 6))

rating_order = df['rating'].value_counts().index

sns.countplot(
    data=df,
    x='rating',
    hue='type',
    order=rating_order
)

plt.title("Netflix Content Ratings by Type", fontsize=16)
plt.xlabel("Content Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.legend(title="Content Type")
plt.tight_layout()

plt.show()
```

## 7. Combined Netflix Content Analysis Dashboard

A combined dashboard is created using Matplotlib and Seaborn to present multiple visualizations in a single figure.

The dashboard includes:

- Movies vs TV Shows comparison.
- Content type percentage distribution.
- Content count visualization.
- Ratings comparison by content type.

This dashboard provides a clear overview of Netflix content distribution.

```python
fig, axes = plt.subplots(2, 2, figsize=(17, 12))

# 1. Professional Bar Chart
bars = axes[0, 0].bar(
    content_counts.index,
    content_counts.values,
    color=['red', 'blue'],
    width=0.5
)
axes[0, 0].bar_label(bars, padding=3)
axes[0, 0].set_title("Movies vs TV Shows", fontsize=14, fontweight='bold')
axes[0, 0].set_ylabel("Number of Titles")
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.4)

# 2. Professional Pie Chart
axes[0, 1].pie(
    content_counts.values,
    labels=content_counts.index,
    autopct='%1.2f%%',
    startangle=90,
    explode=(0.05, 0),
    shadow=True,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)
axes[0, 1].set_title("Content Type Distribution", fontsize=14, fontweight='bold')

# 3. Seaborn Count Plot
ax = sns.countplot(data=df, x='type', ax=axes[1, 0])
for container in ax.containers:
    ax.bar_label(container, padding=3)
axes[1, 0].set_title("Content Type Count", fontsize=14, fontweight='bold')
axes[1, 0].set_ylabel("Number of Titles")

# 4. Ratings by Content Type
sns.countplot(
    data=df,
    x='rating',
    hue='type',
    order=rating_order,
    ax=axes[1, 1]
)
axes[1, 1].set_title("Ratings by Content Type", fontsize=14, fontweight='bold')
axes[1, 1].tick_params(axis='x', rotation=45)
axes[1, 1].set_ylabel("Number of Titles")

plt.suptitle(
    "Netflix Content Type Analysis Dashboard",
    fontsize=19,
    fontweight='bold'
)

plt.tight_layout()
plt.savefig(
    "output/Netflix_Content_Dashboard.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

## 8  . Rating Percentage Analysis

The percentage distribution of content ratings is calculated separately for Movies and TV Shows.

This allows a fair comparison between both content types despite the difference in their total number of titles.

A grouped bar chart is also created to visualize the percentage distribution of different ratings.

```python
rating_percentage = pd.crosstab(
    df['rating'],
    df['type'],
    normalize='columns'
) * 100

rating_percentage.round(2)
```
